````python
# Fix the tab in column name
df.rename(columns={'Daytime/evening attendance\t': 'Daytime/evening attendance'}, inplace=True)

# Map remaining binary/numeric categoricals to readable labels
binary_map = {1: 'Yes', 0: 'No'}
gender_map = {1: 'Male', 0: 'Female'}
attendance_map = {1: 'Daytime', 0: 'Evening'}
marital_map = {1: 'Single', 2: 'Married', 3: 'Widower', 4: 'Divorced', 5: 'Facto union', 6: 'Legally separated'}

df['Gender'] = df['Gender'].map(gender_map)
df['Daytime/evening attendance'] = df['Daytime/evening attendance'].map(attendance_map)
df['Displaced'] = df['Displaced'].map(binary_map)
df['Educational special needs'] = df['Educational special needs'].map(binary_map)
df['Debtor'] = df['Debtor'].map(binary_map)
df['Tuition fees up to date'] = df['Tuition fees up to date'].map(binary_map)
df['Scholarship holder'] = df['Scholarship holder'].map(binary_map)
df['International'] = df['International'].map(binary_map)
df['Marital status'] = df['Marital status'].map(marital_map)

# Define variable groups
cat_vars = [
    'Marital status', 'Application mode', 'Course', 'Daytime/evening attendance',
    'Previous qualification', 'Nacionality', "Mother's qualification", "Father's qualification",
    "Mother's occupation", "Father's occupation", 'Gender', 'Displaced',
    'Educational special needs', 'Debtor', 'Tuition fees up to date',
    'Scholarship holder', 'International'
]

num_continuous = [
    'Previous qualification (grade)', 'Admission grade', 'Age at enrollment',
    'Curricular units 1st sem (grade)', 'Curricular units 2nd sem (grade)',
    'Unemployment rate', 'Inflation rate', 'GDP'
]

num_count = [
    'Application order',
    'Curricular units 1st sem (credited)', 'Curricular units 1st sem (enrolled)',
    'Curricular units 1st sem (evaluations)', 'Curricular units 1st sem (approved)',
    'Curricular units 1st sem (without evaluations)',
    'Curricular units 2nd sem (credited)', 'Curricular units 2nd sem (enrolled)',
    'Curricular units 2nd sem (evaluations)', 'Curricular units 2nd sem (approved)',
    'Curricular units 2nd sem (without evaluations)'
]

PALETTE = {'Graduate': '#2ecc71', 'Dropout': '#e74c3c', 'Enrolled': '#3498db'}
TARGET_ORDER = ['Graduate', 'Enrolled', 'Dropout']

print(f"Dataset: {df.shape[0]:,} rows × {df.shape[1]} columns")
print(f"Categorical: {len(cat_vars)} | Continuous numeric: {len(num_continuous)} | Count numeric: {len(num_count)}")
````

## 3. UNIVARIATE ANALYSIS
### 3.1 Target Distribution

````python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

target_counts = df['Target'].value_counts()[TARGET_ORDER]
target_pct = target_counts / target_counts.sum() * 100

bars = axes[0].bar(TARGET_ORDER, target_counts.values,
                   color=[PALETTE[t] for t in TARGET_ORDER], width=0.5, edgecolor='white')
for bar, val, pct in zip(bars, target_counts.values, target_pct.values):
    axes[0].text(bar.get_x() + bar.get_width()/2, bar.get_height() + 30,
                 f'{val:,}\n({pct:.1f}%)', ha='center', va='bottom', fontsize=11, fontweight='bold')
axes[0].set_title('Target Class Distribution', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Target'); axes[0].set_ylabel('Count')
axes[0].set_ylim(0, target_counts.max() * 1.18)
axes[0].spines[['top', 'right']].set_visible(False)

wedges, texts, autotexts = axes[1].pie(
    target_counts.values,
    labels=TARGET_ORDER,
    autopct='%1.1f%%',
    colors=[PALETTE[t] for t in TARGET_ORDER],
    startangle=90,
    wedgeprops={'edgecolor': 'white', 'linewidth': 2}
)
for autotext in autotexts:
    autotext.set_fontsize(11); autotext.set_fontweight('bold')
axes[1].set_title('Target Class Proportion', fontsize=14, fontweight='bold')

plt.suptitle('Student Academic Outcome Overview\n4,424 Students | 3 Classes', fontsize=13, y=1.01)
plt.tight_layout()
plt.show()

print("\n>> INSIGHT: Dataset is moderately imbalanced — Graduate (50%) dominates. "
      "Dropout (32%) is the main concern for prediction. "
      "Enrolled (18%) are still active students — consider binary framing (Dropout vs. non-Dropout) in modeling.")
````

## 4. BIVARIATE ANALYSIS — Categorical vs. Target
### 4.1 Key Binary Features vs. Target

````python
key_binary = ['Gender', 'Scholarship holder', 'Debtor', 'Tuition fees up to date',
              'Displaced', 'Daytime/evening attendance', 'International']

fig, axes = plt.subplots(2, 4, figsize=(20, 10))
axes = axes.flatten()

for i, col in enumerate(key_binary):
    ct = df.groupby([col, 'Target']).size().unstack(fill_value=0)
    ct_pct = ct.div(ct.sum(axis=1), axis=0) * 100
    ct_pct[TARGET_ORDER].plot(kind='bar', ax=axes[i], color=[PALETTE[t] for t in TARGET_ORDER],
                               width=0.6, edgecolor='white')
    axes[i].set_title(col, fontsize=10, fontweight='bold')
    axes[i].set_xlabel('')
    axes[i].set_ylabel('% of Group')
    axes[i].set_ylim(0, 100)
    axes[i].tick_params(axis='x', rotation=0)
    axes[i].legend(loc='upper right', fontsize=7)
    axes[i].spines[['top', 'right']].set_visible(False)
    
    # Add dropout rate annotation
    for cat in ct_pct.index:
        do_rate = ct_pct.loc[cat, 'Dropout'] if 'Dropout' in ct_pct.columns else 0
        axes[i].annotate(f'DO:{do_rate:.0f}%', xy=(list(ct_pct.index).index(cat), 5),
                         ha='center', fontsize=7.5, color='white', fontweight='bold')

axes[-1].set_visible(False)
plt.suptitle('Target Distribution (%) by Key Binary Features', fontsize=13, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()

print("\n>> KEY INSIGHTS:")
for col in ['Tuition fees up to date', 'Scholarship holder', 'Debtor']:
    ct = df.groupby([col, 'Target']).size().unstack(fill_value=0)
    ct_pct = ct.div(ct.sum(axis=1), axis=0) * 100
    if 'Dropout' in ct_pct.columns:
        for cat in ct_pct.index:
            print(f"  {col} = {cat}: Dropout rate = {ct_pct.loc[cat,'Dropout']:.1f}%")
````

### 4.2 Dropout Rate by Course

````python
course_target = df.groupby(['Course', 'Target']).size().unstack(fill_value=0)
course_target_pct = course_target.div(course_target.sum(axis=1), axis=0) * 100
course_target_pct['Total'] = course_target.sum(axis=1)
course_target_pct = course_target_pct.sort_values('Dropout', ascending=True)

fig, ax = plt.subplots(figsize=(14, 8))
courses = course_target_pct.index.tolist()
x = np.arange(len(courses))

bars_g = ax.barh(x, course_target_pct['Graduate'], color=PALETTE['Graduate'], label='Graduate')
bars_e = ax.barh(x, course_target_pct.get('Enrolled', 0),
                  left=course_target_pct['Graduate'], color=PALETTE['Enrolled'], label='Enrolled')
bars_d = ax.barh(x, course_target_pct.get('Dropout', 0),
                  left=course_target_pct['Graduate'] + course_target_pct.get('Enrolled', 0),
                  color=PALETTE['Dropout'], label='Dropout')

for i, (course, row) in enumerate(course_target_pct.iterrows()):
    do_val = row.get('Dropout', 0)
    n = int(row['Total'])
    ax.text(101, i, f"{do_val:.0f}% (n={n})", va='center', fontsize=8)

ax.set_yticks(x); ax.set_yticklabels(courses, fontsize=9)
ax.set_xlim(0, 120)
ax.set_xlabel('Percentage (%)')
ax.set_title('Target Outcome by Course (sorted by Dropout Rate ↑)', fontsize=13, fontweight='bold')
ax.legend(loc='lower right')
ax.spines[['top', 'right']].set_visible(False)
plt.tight_layout()
plt.show()

print("\n>> INSIGHT: Courses with highest dropout rates are prime candidates for early-warning systems.")
print("   Courses with very low dropout rates may have retention support worth replicating.")
````

### 4.3 Dropout Rate by Application Mode

### 7.2 Point-Biserial Correlation — Each Feature vs. Dropout (Binary Target)

````python
from scipy.stats import pointbiserialr

df['Is_Dropout'] = (df['Target'] == 'Dropout').astype(int)

pb_results = {}
for col in numeric_cols:
    valid = df[[col, 'Is_Dropout']].dropna()
    r, p = pointbiserialr(valid['Is_Dropout'], valid[col])
    pb_results[col] = {'r': r, 'p': p}

pb_df = pd.DataFrame(pb_results).T.sort_values('r')
pb_df['significant'] = pb_df['p'] < 0.05

fig, ax = plt.subplots(figsize=(10, 10))
colors = ['#e74c3c' if r > 0 else '#2ecc71' for r in pb_df['r']]
bars = ax.barh(range(len(pb_df)), pb_df['r'].values, color=colors, edgecolor='white')
ax.set_yticks(range(len(pb_df)))
ax.set_yticklabels([c.replace('Curricular units ', '') for c in pb_df.index], fontsize=8)
ax.axvline(0, color='black', linewidth=1)
ax.set_xlabel('Point-Biserial Correlation with Dropout')
ax.set_title('Feature Correlation with Dropout (Red = More Dropout, Green = Less Dropout)',
             fontsize=12, fontweight='bold')
ax.spines[['top', 'right']].set_visible(False)
for i, (r, sig) in enumerate(zip(pb_df['r'], pb_df['significant'])):
    marker = '***' if sig else ''
    ax.text(r + (0.005 if r >= 0 else -0.005), i, f'{r:.2f}{marker}',
            va='center', ha='left' if r >= 0 else 'right', fontsize=7.5)

plt.tight_layout()
plt.show()

print("\n>> TOP FEATURES negatively correlated with Dropout (protective factors):")
top_protective = pb_df[pb_df['r'] < -0.2].sort_values('r')
for feat, row in top_protective.iterrows():
    print(f"   {feat}: r = {row['r']:.3f}")

print("\n>> TOP FEATURES positively correlated with Dropout (risk factors):")
top_risk = pb_df[pb_df['r'] > 0.1].sort_values('r', ascending=False)
for feat, row in top_risk.iterrows():
    print(f"   {feat}: r = {row['r']:.3f}")
````

````python
df['Financial risk score'] = (
    (df['Tuition fees up to date'] == 'No').astype(int) +
    (df['Debtor'] == 'Yes').astype(int) +
    (df['Scholarship holder'] == 'No').astype(int)
)

ct = df.groupby(['Financial risk score', 'Target']).size().unstack(fill_value=0)
ct_pct = ct.div(ct.sum(axis=1), axis=0) * 100
ct_pct['Total'] = ct.sum(axis=1)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

ct_pct[TARGET_ORDER].plot(kind='bar', ax=axes[0],
                          color=[PALETTE[t] for t in TARGET_ORDER], width=0.6, edgecolor='white')
axes[0].set_title('Target Distribution by Financial Risk Score (%)', fontsize=11, fontweight='bold')
axes[0].set_xlabel('Financial Risk Score (0=low, 3=high)')
axes[0].set_ylabel('% of Group'); axes[0].set_ylim(0, 100)
axes[0].tick_params(axis='x', rotation=0)
axes[0].legend()
axes[0].spines[['top', 'right']].set_visible(False)

for score in ct_pct.index:
    do_r = ct_pct.loc[score, 'Dropout'] if 'Dropout' in ct_pct.columns else 0
    axes[0].annotate(f'DO:{do_r:.0f}%', xy=(score, 5),
                     ha='center', fontsize=9, color='white', fontweight='bold')

axes[1].bar(ct_pct.index, ct_pct['Total'], color='#8e44ad', edgecolor='white')
axes[1].set_title('Sample Count per Financial Risk Score', fontsize=11, fontweight='bold')
axes[1].set_xlabel('Financial Risk Score')
axes[1].set_ylabel('Count')
axes[1].spines[['top', 'right']].set_visible(False)

plt.suptitle('Financial Risk Profile vs. Academic Outcome', fontsize=13, fontweight='bold', y=1.01)
plt.tight_layout()
plt.show()

print("\n>> CRITICAL INSIGHT: Financial risk score = 3 (all 3 risk factors present) → highest dropout rate.")
print("   Tuition not up to date is the single biggest financial signal for dropout prediction.")
print("   Scholarship holders have dramatically better retention — protective financial factor.")
````

### 8.2 Previous Qualification vs. Target

````python
ct = df.groupby(['Previous qualification', 'Target']).size().unstack(fill_value=0)
ct_pct = ct.div(ct.sum(axis=1), axis=0) * 100
ct_pct['Total'] = ct.sum(axis=1)
ct_pct = ct_pct[ct_pct['Total'] >= 20].sort_values('Dropout', ascending=False)

fig, ax = plt.subplots(figsize=(14, 6))
do_vals = ct_pct.get('Dropout', pd.Series(0, index=ct_pct.index))
colors = ['#e74c3c' if d >= do_vals.mean() else '#3498db' for d in do_vals]
bars = ax.bar(range(len(ct_pct)), do_vals.values, color=colors, edgecolor='white')
ax.axhline(do_vals.mean(), color='navy', linestyle='--', linewidth=1.5,
           label=f'Mean Dropout ({do_vals.mean():.1f}%)')

for i, (val, n) in enumerate(zip(do_vals.values, ct_pct['Total'])):
    ax.text(i, val + 0.5, f'{val:.0f}%\n(n={int(n)})', ha='center', fontsize=7.5)

ax.set_xticks(range(len(ct_pct)))
ax.set_xticklabels([s[:35] + '..' if len(s) > 35 else s for s in ct_pct.index],
                   rotation=40, ha='right', fontsize=8)
ax.set_ylabel('Dropout Rate (%)')
ax.set_title('Dropout Rate by Previous Qualification (min. 20 students)', fontsize=13, fontweight='bold')
ax.legend()
ax.spines[['top', 'right']].set_visible(False)
plt.tight_layout()
plt.show()
````

## 9. SUMMARY DASHBOARD — Key Dropout Signals
### 9.1 Dropout Rate Across All Key Categorical Dimensions

````python
summary_results = {}
for col in ['Gender', 'Scholarship holder', 'Debtor', 'Tuition fees up to date',
            'Daytime/evening attendance', 'Displaced', 'International']:
    ct = df.groupby([col, 'Target']).size().unstack(fill_value=0)
    ct_pct = ct.div(ct.sum(axis=1), axis=0) * 100
    for cat in ct_pct.index:
        do_rate = ct_pct.loc[cat, 'Dropout'] if 'Dropout' in ct_pct.columns else 0
        summary_results[f"{col} = {cat}"] = do_rate

summary_s = pd.Series(summary_results).sort_values(ascending=True)

fig, ax = plt.subplots(figsize=(12, 9))
colors = ['#e74c3c' if v >= 40 else '#f39c12' if v >= 30 else '#2ecc71' for v in summary_s.values]
bars = ax.barh(range(len(summary_s)), summary_s.values, color=colors, edgecolor='white')
ax.set_yticks(range(len(summary_s)))
ax.set_yticklabels(summary_s.index, fontsize=9)
ax.axvline(summary_s.mean(), color='navy', linestyle='--', linewidth=1.5,
           label=f'Overall Dropout Mean ({summary_s.mean():.1f}%)')
for i, val in enumerate(summary_s.values):
    ax.text(val + 0.5, i, f'{val:.1f}%', va='center', fontsize=8)
ax.set_xlabel('Dropout Rate (%)')
ax.set_title('Dropout Rate — All Key Categorical Groups\n(Red ≥40%, Orange 30-40%, Green <30%)',
             fontsize=13, fontweight='bold')
ax.legend()
ax.spines[['top', 'right']].set_visible(False)
plt.tight_layout()
plt.show()
````
